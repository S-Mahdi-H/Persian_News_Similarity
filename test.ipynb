```python

from src.preprocessing.basic_preprocessing import basic_preprocess
from src.preprocessing.morphological_preprocessing import morphological_preprocess
from src.preprocessing.stopword_removal import remove_stopwords


examples = [
    "آمریکاست",
    "آمریکا است",
    "دانشجویان",
    "درختان",
    "سیاسی",
    "نپذیرفت",
    "پذیرفت",
    "نمی‌پذیرد",
    "می‌پذیرد",
    "نخواهد رفت",
    "خواهد رفت",
    "نکرد",
    "کرد",
    "نبود",
    "بود",
    "نمی‌شود",
    "می‌شود",
]


for example in examples:

    basic_text = basic_preprocess(example)

    morphological_tokens = morphological_preprocess(
        basic_text
    )

    final_tokens = remove_stopwords(
        morphological_tokens
    )

    print("\nInput:", example)

    print("Morphological analysis:")

    for token in morphological_tokens:
        print(
            f"Surface: {token['surface']} | "
            f"Lemma: {token['lemma']} | "
            f"POS: {token['upos']}"
        )

    print("Final:", " ".join(final_tokens))

    print("-" * 60)
```
